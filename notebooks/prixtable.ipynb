```python
import pandas as pd

# Prices and costs are an ASSUMPTION: no source in the project carries them. What the regret
# metric needs is a table identical across the strategies it compares -- absolute dollars are
# indicative, only the comparison between strategies carries information.
#
# Not fetched from FAOSTAT: its Prices bulk file has no "World" aggregate at all (182 country
# zones, no regional or global rows), because averaging producer prices across countries has
# no meaning without volume weighting.
prix = pd.read_csv("cypd_prix_couts_estimes.csv")

# The label has to match the model's Item categories character for character, or a merge
# drops the crop silently: "Rice paddy" here against "Rice, paddy" in the yield data.
prix["crop"] = prix["crop"].replace({"Rice paddy": "Rice, paddy"})

desaccord = set(pd.read_csv("df.csv")["Item"].unique()) ^ set(prix["crop"])
assert not desaccord, f"libellés désalignés avec le modèle : {desaccord}"

prix
```

```python
# Exported for ML CYPD.ipynb, where the regret metric consumes it. The label fix above is
# the only transformation, and keeping it here leaves a single place responsible for it.
prix.to_csv("prix_couts.csv", index=False)
print(f"{len(prix)} cultures -> prix_couts.csv")
```
